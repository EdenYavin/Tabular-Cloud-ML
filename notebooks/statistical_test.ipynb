```python

def paired_t_test(with_cloud, no_cloud):
    mean_cloud = np.mean(with_cloud)
    mean_no_cloud = np.mean(no_cloud)
    t_stat, p_value = ttest_rel(with_cloud, no_cloud)

    print(f"  Cloud mean:    {mean_cloud:.4f}")
    print(f"  No cloud mean: {mean_no_cloud:.4f}")
    print(f"  t-statistic:   {t_stat:.4f}")
    print(f"  p-value:       {p_value:.6f}")

    if p_value < 0.05:
        winner = "CLOUD" if mean_cloud > mean_no_cloud else "NO CLOUD"
        diff = abs(mean_cloud - mean_no_cloud)
        print(f"  >>> {winner} is significantly better (p={p_value:.4f}, diff={diff:.4f}) <<<")
    else:
        print(f"  No significant difference (p={p_value:.4f})")
```

```python
print("=== ACCURACY ===\n")
for dataset, no_cloud, with_cloud in zip(
        ["heloc"],
        [heloc_acc_no_cloud],
        [heloc_acc_cloud],
):
    print(f"Dataset: {dataset}")
    paired_t_test(with_cloud, no_cloud)
    print()
```

```python
print("=== AUC ===\n")
for dataset, no_cloud, with_cloud in zip(
        ["heloc"],
        [heloc_auc_no_cloud],
        [heloc_auc_cloud],
):
    print(f"Dataset: {dataset}")
    paired_t_test(with_cloud, no_cloud)
    print()
```
