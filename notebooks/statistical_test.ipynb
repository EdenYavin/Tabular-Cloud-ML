```python
datasets = {
    "heloc":    {"acc": (heloc_acc_cloud, heloc_acc_no_cloud),       "auc": (heloc_auc_cloud, heloc_auc_no_cloud)},
    "ring":     {"acc": (ring_acc_cloud, ring_acc_no_cloud),         "auc": (ring_auc_cloud, ring_auc_no_cloud)},
    "magic":    {"acc": (magic_acc_cloud, magic_acc_no_cloud),       "auc": (magic_auc_cloud, magic_auc_no_cloud)},
    "twonorm":  {"acc": (twonorm_acc_cloud, twonorm_acc_no_cloud),   "auc": (twonorm_auc_cloud, twonorm_auc_no_cloud)},
    "mushroom": {"acc": (mushroom_acc_cloud, mushroom_acc_no_cloud), "auc": (mushroom_auc_cloud, mushroom_auc_no_cloud)},
}

rows = []
for name, metrics in datasets.items():
    for metric, (cloud, no_cloud) in metrics.items():
        t_stat, p_value = ttest_rel(cloud, no_cloud)
        mean_cloud = np.mean(cloud)
        mean_no_cloud = np.mean(no_cloud)
        significant = p_value < 0.05
        winner = ""
        if significant:
            winner = "cloud" if mean_cloud > mean_no_cloud else "no_cloud"
        rows.append({
            "dataset": name,
            "metric": metric,
            "cloud_mean": round(mean_cloud, 4),
            "no_cloud_mean": round(mean_no_cloud, 4),
            "diff": round(mean_cloud - mean_no_cloud, 4),
            "t_statistic": round(t_stat, 4),
            "p_value": round(p_value, 6),
            "significant": significant,
            "winner": winner,
        })

df = pd.DataFrame(rows)
df.to_csv("statistical_tests.csv", index=False)
df
```
